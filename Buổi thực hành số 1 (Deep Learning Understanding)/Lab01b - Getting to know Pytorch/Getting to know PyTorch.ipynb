### **Bài 6. Phép toán và nhân ma trận**
Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.

**Yêu cầu**:
- Phân biệt phép toán a * b với a @ b.T.
- Cho biết shape của kết quả a @ b.T.

```python
a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:", a)
print("b:", b)
print("a + b:", a + b)
print("a / b:", a / b)
print("a^2:", a ** 2)
print("Matrix multiplication:", a @ b.T)
```

### **Bài 7. Các hàm thống kê trên Tensor**
Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

**Yêu cầu**: Thay dim=0 bằng dim=1 và nhận xét sự khác nhau.

```python
a = torch.randn(3, 4)

print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("Standard deviation:")
print(torch.std(a, dim=0))
print("Cumulative sum:")
print(torch.cumsum(a, dim=0))
```

```python
# dim=1 tính theo từng hàng
# dim=0 tính theo từng cột
print("Mean theo hàng (dim=1):", torch.mean(a, dim=1))
print("Standard deviation theo hàng (dim=1):", torch.std(a, dim=1))
print("Cumulative sum theo hàng (dim=1):", torch.cumsum(a, dim=1))
```

### **Nhận xét bài 7**
Sự khác nhau giữa `dim=0` và `dim=1`:
- `dim=0`: phép tính được thực hiện theo từng cột nên kết quả của `mean` và `std` có 4 phần tử; `cumsum` cộng dồn theo chiều từ trên xuống dưới.
- `dim=1`: phép tính được thực hiện theo từng hàng nên `mean` và `std` có 3 phần tử; `cumsum` cộng dồn từ trái sang phải trong mỗi hàng.

### **Bài 8. Autograd cơ bản**
Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

**Yêu cầu**:
- Tính thủ công đạo hàm của y = x² tại x = 3.6 và so sánh với x.grad.
- Đổi hàm thành y = 3x² + 2x + 1 và kiểm tra gradient.

```python
x = torch.tensor(
    3.6,
    requires_grad=True
)

y = x * x
y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

```python
# Với y = x^2
print("Đạo hàm thủ công của y = x^2 tại x = 3.6:", 2 * x.item())
print("Gradient do autograd tính:", x.grad.item())

# Tạo một Tensor mới để gradient của hàm mới không cộng dồn với gradient cũ.
x_new = torch.tensor(3.6, requires_grad=True)
y_new = 3 * x_new**2 + 2 * x_new + 1
y_new.backward()

# Với y = 3x^2 + 2x + 1
print("Đạo hàm thủ công của hàm mới:", 6 * x_new.item() + 2)
print("Gradient của hàm mới:", x_new.grad.item())
```

### **Nhận xét bài 8**

Với $y=x^2$, ta có $\frac{dy}{dx}=2x$, nên tại $x=3.6$ ta sẽ có đạo hàm bằng $7.2$. Kết quả của `x.grad` trùng với phép tính thủ công.

Với $y=3x^2+2x+1$, đạo hàm là $6x+2$ và tại $x=3.6$ nên ta sẽ có đạo hàm bằng $23.6$, cũng trùng với kết quả của autograd.

### **Bài 9. Gradient Descent tìm hệ số đa thức**
Sinh dữ liệu từ đa thức y = x² + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

**Yêu cầu**:
- So sánh ba hệ số học được với [1, 2, 3].
- Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().

```python
polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
    
print("Learned coefficients:")
print(w)
```

```python
# Các hệ số kỳ vọng là [1, 2, 3], tương ứng với x^2, x và hằng số.
he_so_hoc_duoc = w.detach().flatten()
he_so_mong_doi = torch.tensor([1.0, 2.0, 3.0])
print("Hệ số kỳ vọng:", he_so_mong_doi)
print("Hệ số học được:", he_so_hoc_duoc)
print("Sai số tuyệt đối từng hệ số:", torch.abs(he_so_hoc_duoc - he_so_mong_doi))
```

### **Nhận xét bài 9**

Các hệ số học được gần với `[1, 2, 3]`, tương ứng với đa thức $x^2+2x+3$; sai khác rất nhỏ do tính toán số thực và quá trình tối ưu. Trong mỗi vòng lặp:
- `optimizer.zero_grad()`: xóa gradient của vòng lặp trước để gradient không bị cộng dồn.
- `mse.backward()`: dùng autograd để tính gradient của hàm mất mát theo các hệ số.
- `optimizer.step()`: cập nhật các hệ số theo gradient và learning rate của NAdam nhằm giảm MSE.

### **Bài 10. Giải hệ phương trình bằng Autograd**
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

**Yêu cầu**:
- Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.
- Nhận xét vì sao bài toán có thể có nhiều nghiệm.

```python
# Cố định seed để nghiệm số có thể tái lập giữa các lần chạy.
random.seed(0)
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

```python
# Kiểm tra trực tiếp bốn vế trái của hệ phương trình sau khi tối ưu.
sai_so_he = torch.stack([
    A + B - 9,
    C - D - 1,
    A + C - 8,
    B - D - 2,
])
print("Sai số của bốn phương trình:", sai_so_he.detach())
print("Sai số tuyệt đối lớn nhất:", sai_so_he.abs().max().item())
print("A + B =", (A + B).item(), "| cần bằng 9")
print("C - D =", (C - D).item(), "| cần bằng 1")
print("A + C =", (A + C).item(), "| cần bằng 8")
print("B - D =", (B - D).item(), "| cần bằng 2")
```

### **Nhận xét bài 10**

Các giá trị tìm được kiểm tra đúng bốn phương trình với sai số rất nhỏ do tối ưu số. Hệ không có nghiệm duy nhất vì chỉ có ba phương trình độc lập. Đặt $A=t$ thì:

$$
B=9-t, \qquad C=8-t, \qquad D=7-t.
$$

Vì $t$ có thể nhận nhiều giá trị, hệ có vô số nghiệm.

### **Bài 11. Viết chương trình PyTorch để thực hiện các yêu cầu sau**
- Tạo Tensor X có 24 phần tử và reshape lần lượt thành 4 × 6, 2 × 12 và 2 × 3 × 4.
- Tạo hai Tensor A, B kích thước 3 × 4; tính A+B, A*B, A/B, trung bình theo cột và độ lệch chuẩn theo cột.
- Cho y = 2x² + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.
- Thay đa thức trong Bài 9 bằng y = 2x² - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
# Bài 11.1: Reshape giữ nguyên 24 phần tử.
X = torch.arange(24)
print("X ban đầu:", X)
for shape in [(4, 6), (2, 12), (2, 3, 4)]:
    reshaped_X = X.reshape(shape)
    print(f"X reshape {shape}: shape={reshaped_X.shape}, numel={reshaped_X.numel()}")

# Bài 11.2: A * B là phép nhân từng phần tử; mean/std theo dim=0 là theo cột.
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("A + B:\n", A + B)
print("A * B:\n", A * B)
print("A / B:\n", A / B)
print("Trung bình theo cột:\n", torch.mean(A, dim=0))
print("Độ lệch chuẩn theo cột:\n", torch.std(A, dim=0))

# Bài 11.3: Autograd cho y = 2x^2 + 5x + 3, ta có dy/dx= 4x+5. Tại x=2, đạo hàm sẽ bằng 13.
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()
thucong = 4 * x.item() + 5
print("dy/dx bằng autograd:", x.grad.item())
print("dy/dx thủ công = 4x + 5:", thucong)

# Bài 11.4: Dùng mô hình tuyến tính theo các đặc trưng [x^2, x, 1]
# Học ba hệ số của đa thức y = 2x^2 - 3x + 5.
torch.manual_seed(0)
X_poly = torch.linspace(-5, 5, 40).reshape(-1, 1)
Y_poly = 2 * X_poly**2 - 3 * X_poly + 5
features = torch.cat([X_poly**2, X_poly, torch.ones_like(X_poly)], dim=1)
coefficients = torch.randn(3, 1, requires_grad=True)
poly_optimizer = torch.optim.NAdam([coefficients], lr=0.01)

# Tăng số bước để NAdam hội tụ đủ ở cả hệ số x^2 và hệ số tự do.
for _ in range(5000):
    poly_optimizer.zero_grad()
    predictions = features @ coefficients
    loss = torch.mean((predictions - Y_poly) ** 2)
    loss.backward()
    poly_optimizer.step()

print("Hệ số học được [x^2, x, 1]:", coefficients.detach().flatten())
print("Hệ số mong đợi [2, -3, 5]:", torch.tensor([2.0, -3.0, 5.0]))
print("MSE cuối:", loss.item())
```

### **Nhận xét bài 11.4**

Mô hình dùng ba đặc trưng $[x^2,x,1]$, nên ba hệ số học được lần lượt gần với $[2,-3,5]$. MSE gần 0 cho thấy đa thức đã được học đúng.
